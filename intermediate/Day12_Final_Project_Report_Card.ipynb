````python
import pandas as pd
````

````python
import matplotlib.pyplot as plt
````

### Step 1: Student data

````python
data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "Anu", "Bikash"],
    "Math": [78, 92, 65, 88, 34, 71],
    "Science": [82, 89, 70, 91, 45, 66],
    "English": [70, 85, 72, 80, 50, 74]
}
````

````python
df = pd.DataFrame(data)
````

````python
df
````

### Step 2: Total and percentage

````python
df["Total"] = df["Math"] + df["Science"] + df["English"]
````

````python
df["Percentage"] = round(df["Total"] / 3, 1)
````

### Step 3: A function for the grade
`.apply(grade)` runs our function for **every** row.

````python
def grade(p):
    if p >= 90:
        return "A+"
    elif p >= 80:
        return "A"
    elif p >= 60:
        return "B"
    elif p >= 40:
        return "C"
    else:
        return "Fail"
````

````python
df["Grade"] = df["Percentage"].apply(grade)
````

````python
df
````

### Step 4: Rank and topper

````python
df = df.sort_values("Percentage", ascending=False)
````

````python
topper = df.iloc[0]   # first row after sorting
````

````python
print(f"🥇 Topper: {topper['Name']} with {topper['Percentage']}%")
````

### Step 5: Save the result

````python
df.to_csv("result.csv", index=False)
````

### Step 6: Result chart

````python
plt.bar(df["Name"], df["Percentage"], color="teal")
plt.title("Class Result")
plt.ylabel("Percentage")
plt.show()
````

### Step 7: Search a student

````python
search = input("Enter a student name: ")
````

````python
found = df[df["Name"] == search]
````

````python
if len(found) == 0:
    print("Student not found")
else:
    print(found)
````
