# Day 2: Functions 🧩

**Today you will learn:**
- Making your own commands with `def`
- Giving data to a function (parameters)

### 1. Why functions?
A function is a **set of steps with a name**. Write once, use many times.

```python
def say_hello():
    print("Hello!")
    print("Welcome to Python class")
```

Making the function does nothing yet. We have to **call** it:

```python
say_hello()
```

```python
say_hello()
```

### 2. Parameters
Parameters are inputs for the function. They go inside `( )`.

```python
def greet(name):
    print(f"Namaste, {name}!")
```

```python
greet("Ram")
```

```python
greet("Sita")
```

### 3. More than one parameter

```python
def add(a, b):
    print(a + b)
```

```python
add(5, 3)
```

```python
add(100, 250)
```

### 4. Functions with loops

```python
def table(n):
    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")
```

```python
table(3)
```

### 5. Functions with if

```python
def check_even(n):
    if n % 2 == 0:
        print(n, "is even")
    else:
        print(n, "is odd")
```

```python
check_even(10)
```

```python
check_even(7)
```

## 📌 Quick Recap
- `def name():` makes a function. `name()` calls it.
- Values inside `( )` are **parameters**.
- A function can have loops and `if` inside it.
