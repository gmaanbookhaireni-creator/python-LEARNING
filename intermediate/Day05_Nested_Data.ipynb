# Day 5: Nested Data and Nested Loops 🪆

**Today you will learn:**
- Lists inside lists
- A list of dictionaries (like a class register)
- A loop inside a loop

### 1. List of lists
Like a table with rows and columns.

````python
marks = [
    [80, 75, 90],
    [60, 85, 70],
    [95, 88, 92]
]
````

````python
print(marks[0])
````

````python
print(marks[0][2])
````

### 2. A loop inside a loop
The inner loop runs fully for **each** round of the outer loop.
`print(m, end=" ")` keeps printing on the same line.

````python
for row in marks:
    for m in row:
        print(m, end=" ")
    print()
````

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
    print("---")
````

### 3. A list of dictionaries
Each dictionary is one student.

````python
students = [
    {"name": "Ram", "marks": 78},
    {"name": "Sita", "marks": 92},
    {"name": "Hari", "marks": 65}
]
````

````python
print(students[1]["name"])
````

````python
for s in students:
    print(f"{s['name']} got {s['marks']}")
````

### 4. Find the topper

````python
topper = students[0]
for s in students:
    if s["marks"] > topper["marks"]:
        topper = s
print("Topper:", topper["name"])
````

### 5. Add a new student

````python
students.append({"name": "Gita", "marks": 88})
````

````python
print(len(students))
````

## 📌 Quick Recap
- `table[row][column]` gets one item from a list of lists.
- A loop inside a loop → the inner loop runs again and again.
- A list of dictionaries is like a class register.
