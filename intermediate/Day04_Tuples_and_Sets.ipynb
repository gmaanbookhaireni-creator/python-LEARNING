# Day 4: Tuples and Sets 📦

**Today you will learn:**
- **Tuples**: lists that cannot change
- **Sets**: collections with no repeated items

### 1. Tuples
Use round brackets `( )`. You can read items, but you cannot change them.

```python
colors = ("red", "green", "blue")
```

```python
print(colors[0])
```

```python
print(len(colors))
```

Run this. Tuples cannot be changed, so we get an error:

```python
colors[0] = "pink"
```

### 2. Unpacking a tuple
Put each item into its own variable.

```python
point = (3, 5)
```

```python
x, y = point
```

```python
print("x =", x, "y =", y)
```

### 3. Sets
Use curly brackets `{ }`. A set **removes repeated items**. The order is not fixed.

```python
numbers = {1, 2, 2, 3, 3, 3}
```

```python
print(numbers)
```

```python
numbers.add(10)
```

```python
numbers.remove(1)
```

```python
print(numbers)
```

### 4. Remove repeats from a list

```python
names = ["Ram", "Sita", "Ram", "Hari", "Sita"]
```

```python
unique = set(names)
```

```python
print(unique)
```

### 5. Joining sets
- `a | b` → items in **either** set (union)
- `a & b` → items in **both** sets (intersection)

```python
class_a = {"Ram", "Sita", "Hari"}
```

```python
class_b = {"Sita", "Gita", "Hari"}
```

```python
print(class_a | class_b)
```

```python
print(class_a & class_b)
```

### List vs Tuple vs Set
| | List `[ ]` | Tuple `( )` | Set `{ }` |
|---|---|---|---|
| Can change? | ✅ | ❌ | ✅ |
| Repeats allowed? | ✅ | ✅ | ❌ |
| Has order / index? | ✅ | ✅ | ❌ |

## 📌 Quick Recap
- Tuple `( )`: fixed, cannot change.
- Set `{ }`: no repeats, `add`, `remove`.
- `|` union, `&` intersection.
