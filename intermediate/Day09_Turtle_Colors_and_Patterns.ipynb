# Day 9: Turtle Colours and Patterns 🎨

**Today you will learn:**
- Colours, pen size and filling shapes
- Making patterns with loops and functions

```python
import turtle
```

```python
t = turtle.Turtle()
```

```python
screen = turtle.Screen()
```

### 1. Colours and pen size

```python
screen.bgcolor("lightyellow")
```

```python
t.pencolor("blue")
```

```python
t.pensize(5)
```

```python
t.forward(150)
```

### 2. Fill a shape
Write `begin_fill()` before drawing and `end_fill()` after.

```python
t.reset()
```

```python
t.fillcolor("orange")
```

```python
t.begin_fill()
for i in range(4):
    t.forward(100)
    t.right(90)
t.end_fill()
```

### 3. Functions for shapes
Make a shape once, draw it anywhere.

```python
def square(size, color):
    t.fillcolor(color)
    t.begin_fill()
    for i in range(4):
        t.forward(size)
        t.right(90)
    t.end_fill()
```

```python
t.reset()
```

```python
square(50, "red")
```

```python
t.penup()
```

```python
t.goto(100, 0)
```

```python
t.pendown()
```

```python
square(80, "green")
```

### 4. Spiral
Each line is a bit longer than the last one.

```python
t.reset()
```

```python
t.speed(0)
```

```python
for i in range(50):
    t.forward(i * 5)
    t.right(90)
```

### 5. Flower
Draw many circles, turning a little each time.

```python
t.reset()
```

```python
t.speed(0)
```

```python
colors = ["red", "orange", "yellow", "green", "blue", "purple"]
```

```python
for i in range(36):
    t.pencolor(colors[i % 6])
    t.circle(80)
    t.right(10)
```

## 📌 Quick Recap
- `pencolor`, `fillcolor`, `pensize`, `screen.bgcolor`
- `begin_fill()` … draw … `end_fill()`
- Loops + small changes = patterns!
