### 1. Variables and f-strings

```python
name = "Sita"
```

```python
age = 15
```

```python
print(f"{name} is {age} years old.")
```

### 2. if / elif / else

```python
marks = 72
```

```python
if marks >= 80:
    print("Distinction")
elif marks >= 40:
    print("Pass")
else:
    print("Fail")
```

### 3. for loop

```python
total = 0
for i in range(1, 6):
    total += i
print("Sum:", total)
```

### 4. while loop

```python
count = 3
while count > 0:
    print(count)
    count -= 1
print("Go!")
```

### 5. Lists

```python
fruits = ["apple", "mango", "banana"]
```

```python
fruits.append("orange")
```

```python
print(fruits)
```

```python
print(len(fruits))
```

### 6. Dictionaries

```python
student = {"name": "Ram", "class": 10, "marks": 85}
```

```python
print(student["name"])
```

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

## 📌 Quick Recap
- f-strings: `f"Hello {name}"`
- `if` / `elif` / `else`, `for`, `while`
- Lists `[ ]` and dictionaries `{key: value}`
