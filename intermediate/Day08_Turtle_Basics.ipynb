**Today you will learn:**
- Moving a turtle to draw lines
- Drawing shapes with loops

### 1. Make a turtle

```python
import turtle
```

```python
t = turtle.Turtle()
```

### 2. Move and turn
- `forward(100)` → move 100 steps
- `right(90)` / `left(90)` → turn 90 degrees

```python
t.forward(100)
```

```python
t.right(90)
```

```python
t.forward(100)
```

```python
t.left(45)
```

```python
t.backward(50)
```

### 3. Start again
`t.reset()` clears the drawing and sends the turtle home.

```python
t.reset()
```

### 4. Draw a square with a loop
4 sides, turn 90° each time.

```python
for i in range(4):
    t.forward(100)
    t.right(90)
```

### 5. Triangle
3 sides, turn 120° each time.

```python
t.reset()
```

```python
for i in range(3):
    t.forward(100)
    t.left(120)
```

### 6. Move without drawing

```python
t.penup()
```

```python
t.goto(-150, 100)
```

```python
t.pendown()
```

```python
t.circle(50)
```

### 7. Speed and shape

```python
t.speed(1)      # 1 slow ... 10 fast, 0 fastest
```

```python
t.shape("turtle")
```

## 📌 Quick Recap
- `forward`, `backward`, `right`, `left`, `circle`
- `penup` / `pendown` to move without drawing, `goto(x, y)` to jump
- Shape rule: turn = `360 ÷ number of sides`
