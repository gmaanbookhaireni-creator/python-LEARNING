# Day 3: return, Default Values and Scope ↩️

**Today you will learn:**
- Getting an answer back with `return`
- Default values for parameters
- Local and global variables

### 1. `print` vs `return`
- `print` only **shows** the answer.
- `return` **gives back** the answer, so we can save and use it.

```python
def square(n):
    return n * n
```

```python
result = square(4)
```

```python
print(result)
```

```python
print(square(5) + square(2))
```

### 2. Return text

```python
def grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"
```

```python
print(grade(85))
```

```python
print(grade(35))
```

### 3. Default values
If we don't give a value, the default is used.

```python
def greet(name, message="Good morning"):
    print(f"{message}, {name}!")
```

```python
greet("Ram")
```

```python
greet("Sita", "Good night")
```

### 4. Scope: local and global
A variable made **inside** a function lives only inside it (local).
A variable made **outside** can be read anywhere (global).

```python
school = "Ram Shah School"   # global
```

```python
def show():
    city = "Kathmandu"       # local
    print(school, city)
```

```python
show()
```

Run this. `city` lives only inside `show()`, so we get an error:

```python
print(city)
```

## 📌 Quick Recap
- `return` gives back a value. Save it: `x = square(4)`.
- `def greet(name, message="Hi"):` → `message` has a default.
- Variables made inside a function are **local**.
