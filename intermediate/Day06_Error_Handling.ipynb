| Error | When |
|---|---|
| `ValueError` | wrong kind of value, like `int("abc")` |
| `ZeroDivisionError` | divide by 0 |
| `IndexError` | position not in the list |
| `NameError` | variable does not exist |

### 2. `try` / `except`
Python **tries** the code. If an error happens, it runs `except` instead of crashing.

```python
try:
    age = int(input("Enter your age: "))
    print("Next year you will be", age + 1)
except:
    print("Please type a number!")
```

### 3. Catch a specific error

```python
try:
    a = int(input("First number: "))
    b = int(input("Second number: "))
    print(a / b)
except ValueError:
    print("Only numbers please!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

### 4. Keep asking until the input is correct

```python
while True:
    try:
        marks = int(input("Enter marks: "))
        break
    except ValueError:
        print("That is not a number. Try again.")
```

```python
print("You entered", marks)
```

## 📌 Quick Recap
- Errors stop the program. Read the **last line** to know why.
- `try:` risky code → `except:` what to do if it fails.
- Use `while True` + `try` + `break` to ask until the input is right.
