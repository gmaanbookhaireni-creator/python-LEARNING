# Day 11: Filtering Data and Drawing Charts 📊

**Today you will learn:**
- Filtering and sorting a table
- Drawing bar, line and pie charts with **matplotlib**

> If an import gives an error, run `%pip install pandas matplotlib` in a cell first.

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "Name": ["Ram", "Sita", "Hari", "Gita", "Anu"],
    "Math": [78, 92, 65, 88, 54],
    "Science": [82, 89, 70, 91, 60]
})
```

### 1. Filter rows
Get only the rows that match a condition.

```python
df[df["Math"] > 70]
```

```python
df[df["Name"] == "Gita"]
```

### 2. Sort

```python
df.sort_values("Math")
```

```python
df.sort_values("Math", ascending=False)
```

### 3. Bar chart

```python
plt.bar(df["Name"], df["Math"])
plt.title("Math Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

### 4. Line chart
Good for showing change over time.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
```

```python
temperature = [12, 15, 20, 25, 28]
```

```python
plt.plot(months, temperature, marker="o")
plt.title("Kathmandu Temperature")
plt.show()
```

### 5. Pie chart
Good for showing parts of a whole.

```python
activities = ["School", "Sleep", "Play", "Study"]
```

```python
hours = [7, 8, 4, 5]
```

```python
plt.pie(hours, labels=activities, autopct="%1.0f%%")
plt.title("My Day")
plt.show()
```

## 📌 Quick Recap
- Filter: `df[df["Math"] > 70]`
- Sort: `df.sort_values("Math", ascending=False)`
- Charts: `plt.bar`, `plt.plot`, `plt.pie`, then `plt.show()`
