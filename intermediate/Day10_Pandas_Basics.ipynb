**Today you will learn:**
- Making a table (**DataFrame**)
- Getting columns
- Saving and reading **CSV** files

> If `import pandas` gives an error, run `pip install pandas` in a cell first: `%pip install pandas`

```python
import pandas as pd
```

### 1. Make a table from a dictionary
Each key becomes a **column**.

```python
data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "Anu"],
    "Math": [78, 92, 65, 88, 54],
    "Science": [82, 89, 70, 91, 60],
    "English": [70, 85, 72, 80, 66]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

### 2. Look at the table

```python
df.head(3)      # first 3 rows
```

```python
df.shape        # (rows, columns)
```

### 3. Get columns

```python
df["Math"]
```

```python
df[["Name", "Math"]]
```

### 4. Quick maths on a column

```python
df["Math"].mean()
```

```python
df["Math"].max()
```

```python
df["Science"].min()
```

### 5. Make a new column

```python
df["Total"] = df["Math"] + df["Science"] + df["English"]
```

```python
df
```

### 6. Save to CSV and read it back
A **CSV** file is a simple table that Excel can also open.

```python
df.to_csv("marks.csv", index=False)
```

```python
new_df = pd.read_csv("marks.csv")
```

```python
new_df
```

## 📌 Quick Recap
- `pd.DataFrame(dictionary)` makes a table.
- `df["Column"]` gets one column. `.mean()`, `.max()`, `.min()`.
- `df.to_csv("file.csv", index=False)` saves, `pd.read_csv("file.csv")` reads.
