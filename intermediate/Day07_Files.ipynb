# Day 7: Working with Files 📁

**Today you will learn:**
- Writing to a file
- Reading a file
- Adding (appending) to a file

### File modes
| Mode | Meaning |
|---|---|
| `"w"` | **write**: makes a new file (deletes old content!) |
| `"r"` | **read**: reads the file |
| `"a"` | **append**: adds to the end |

### 1. Write a file
`with open(...)` opens the file and closes it automatically.
`\n` means "new line".

```python
with open("diary.txt", "w") as f:
    f.write("Today I learned Python files.\n")
    f.write("It was fun!\n")
```

Look in the file list on the left. You will see `diary.txt`!

### 2. Read a file

```python
with open("diary.txt", "r") as f:
    content = f.read()
```

```python
print(content)
```

### 3. Append to a file
`"a"` keeps the old lines and adds new ones at the end.

```python
with open("diary.txt", "a") as f:
    f.write("Tomorrow I will learn turtle.\n")
```

```python
with open("diary.txt", "r") as f:
    print(f.read())
```

### 4. Read line by line

```python
with open("diary.txt", "r") as f:
    for line in f:
        print("->", line.strip())
```

### 5. Save user input to a file

```python
name = input("Your name: ")
```

```python
with open("names.txt", "a") as f:
    f.write(name + "\n")
```

### 6. File not found?

```python
try:
    with open("nothing.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found!")
```

## 📌 Quick Recap
- `with open("file.txt", "w") as f:` → `f.write("...")`
- `"r"` read, `"w"` write (new), `"a"` add to the end.
- `\n` makes a new line. `line.strip()` removes it.
