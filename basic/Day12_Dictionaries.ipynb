# Day 12: Dictionaries 📖

**Today you will learn:**
- Storing data as **key : value** pairs
- Getting, changing, adding and removing values
- Looping through a dictionary

### 1. Make a dictionary
Like a real dictionary: look up a **key** to get its **value**. Use curly brackets `{ }`.

```python
student = {
    "name": "Ram",
    "roll": 1,
    "class": 9,
    "is_present": True
}
```

```python
print(student)
```

### 2. Get a value

```python
print(student["name"])
```

```python
print(student["class"])
```

### 3. Change, add and remove

```python
student["class"] = 10          # change
```

```python
student["school"] = "Ram Shah" # add new key
```

```python
del student["is_present"]      # remove
```

```python
print(student)
```

### 4. Keys, values and items

```python
print(student.keys())
```

```python
print(student.values())
```

```python
print(student.items())
```

### 5. Loop through a dictionary

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

### 6. Check if a key exists

```python
if "name" in student:
    print("Name is saved")
```

### 7. Example: report card
A dictionary can have another dictionary inside it.

```python
student = {
    "name": "Roshan",
    "class": 10,
    "marks": {
        "math": 30,
        "science": 50,
        "english": 40,
        "nepali": 80,
        "computer": 88
    }
}
```

```python
total = sum(student["marks"].values())
```

```python
percentage = total / len(student["marks"])
```

```python
print(f"{student['name']} got {total} marks ({percentage}%)")
```

## 📌 Quick Recap
- `d = {"key": value}`, get with `d["key"]`.
- `d["new"] = x` adds, `del d["key"]` removes.
- `keys()`, `values()`, `items()` and `in`.
