# Day 10: Lists 📋

**Today you will learn:**
- Storing many values in one **list**
- Getting, changing, adding and removing items
- Useful list tools

### 1. Make a list
Use square brackets `[ ]` and separate items with commas.

```python
fruits = ["apple", "mango", "banana", "orange"]
```

```python
print(fruits)
```

### 2. Get an item (index)
Counting starts from **0**. `-1` means the last item.

```python
print(fruits[0])
```

```python
print(fruits[2])
```

```python
print(fruits[-1])
```

### 3. Change an item

```python
fruits[1] = "grapes"
```

```python
print(fruits)
```

### 4. Add and remove
| Code | Does |
|---|---|
| `append(x)` | add at the end |
| `insert(i, x)` | add at position `i` |
| `remove(x)` | remove item `x` |

```python
fruits.append("kiwi")
```

```python
print(fruits)
```

```python
fruits.insert(1, "litchi")
```

```python
print(fruits)
```

```python
fruits.remove("banana")
```

```python
print(fruits)
```

### 5. Length and checking with `in`

```python
print(len(fruits))
```

```python
if "apple" in fruits:
    print("Yes, we have apple")
```

### 6. Sort

```python
numbers = [5, 1, 7, 3, 9, 2]
```

```python
numbers.sort()
```

```python
print(numbers)
```

```python
numbers.sort(reverse=True)
```

```python
print(numbers)
```

### 7. Slicing — get a part of the list
`list[start:stop]`. The stop item is **not** included.

```python
numbers = [10, 20, 30, 40, 50, 60]
```

```python
print(numbers[1:4])
```

```python
print(numbers[:3])
```

### 8. Math on a list

```python
marks = [80, 65, 90, 75, 88]
```

```python
print("Total  :", sum(marks))
```

```python
print("Highest:", max(marks))
```

```python
print("Lowest :", min(marks))
```

```python
print("Average:", sum(marks) / len(marks))
```

## 📌 Quick Recap
- `my_list = [a, b, c]`, first item is `my_list[0]`.
- `append`, `insert`, `remove`, `sort`.
- `len`, `sum`, `max`, `min`, `in`.
