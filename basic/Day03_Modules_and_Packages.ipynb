# Day 3: Modules, Packages and pip 📦

**Today you will learn:**
- What modules and packages are
- Different ways to `import`
- Installing new packages with `pip`

### 1. What are they?
| Name | Meaning | Examples |
|---|---|---|
| **Module** | A file full of ready-made tools | `math`, `random`, `time`, `datetime` |
| **Package** | A big box of modules, made by other people | `pandas`, `matplotlib`, `pyjokes` |

Modules like `math` come **with** Python. Most packages must be **installed** first.

### 2. `import module`
Use the tool with the module name in front: `math.sqrt`.

```python
import math
```

```python
print(math.sqrt(25))
```

```python
print(math.pi)
```

### 3. `from module import tool`
Take only one tool. Now you don't need `math.` in front.

```python
from math import sqrt
```

```python
print(sqrt(49))
```

```python
from datetime import date
```

```python
print(date.today())
```

### 4. `import module as nickname`
Give a long name a short nickname.

```python
import random as rd
```

```python
print(rd.randint(1, 6))
```

### 5. Package not installed?
Run this. Python can't find it, so we get an error:

```python
import pyjokez
```

### 6. Install a package with `pip`
`pip` downloads packages from the internet. In a notebook, write `%pip install name`.

```python
%pip install pyjokes art
```

### 7. Use the new packages

```python
import pyjokes
```

```python
print(pyjokes.get_joke(category="neutral"))
```

```python
import art
```

```python
art.tprint("Python")
```

### 8. Ask Python for help
`help()` shows what a tool does.

```python
help(math.sqrt)
```

## 📌 Quick Recap
- **Module** = a file of tools. **Package** = a big box of modules.
- `import math` → `math.sqrt()` · `from math import sqrt` → `sqrt()` · `import random as rd` → `rd.randint()`
- `%pip install name` installs a package. Install once, import every time.
- `ModuleNotFoundError` → the name is wrong or the package is not installed.
