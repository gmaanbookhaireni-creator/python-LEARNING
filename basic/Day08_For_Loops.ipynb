# Day 8: For Loops 🔁

**Today you will learn:**
- Repeating code with `for`
- Counting with `range()`
- Adding numbers inside a loop

### 1. Why loops?
Instead of writing `print()` 5 times, let a loop do it.

````python
for i in range(5):
    print("Hello!")
````

### 2. `range()`
| Code | Gives |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

````python
for i in range(1, 6):
    print(i)
````

````python
for i in range(2, 11, 2):
    print(i)
````

````python
for i in range(5, 0, -1):
    print(i)
print("Blast off! 🚀")
````

### 3. Loop over a list or a word

````python
for name in ["Ram", "Shyam", "Gita"]:
    print(name + " is a good student")
````

````python
for letter in "Nepal":
    print(letter)
````

### 4. Adding inside a loop
Start with `0`, then add each number.

````python
total = 0
for i in range(1, 11):
    total += i
print("Sum of 1 to 10 =", total)
````

### 5. Multiplication table

````python
num = 5
````

````python
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
````

### 6. Star pattern
`"*" * 3` gives `***`.

````python
for i in range(6, 0, -1):
    print("*" * i)
````

## 📌 Quick Recap
- `for i in range(...)` repeats code.
- `range(start, stop, step)`. The stop number is **not** included.
- To add in a loop: `total = 0` first, then `total += i`.
