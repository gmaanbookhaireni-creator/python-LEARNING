# Day 11: Playing with Text (Strings) 🔤

**Today you will learn:**
- Getting letters from text
- Useful string tools
- Splitting and joining text

### 1. Letters have positions too
Just like lists, counting starts from **0**.

```python
word = "Python"
```

```python
print(word[0])
```

```python
print(word[-1])
```

```python
print(len(word))
```

```python
print(word[0:3])
```

Reverse a word with `[::-1]`:

```python
print(word[::-1])
```

### 2. Big and small letters

```python
name = "ram bahadur"
```

```python
print(name.upper())
```

```python
print(name.lower())
```

```python
print(name.title())
```

### 3. Replace and count

```python
sentence = "ram is a good boy, ram likes momo"
```

```python
print(sentence.replace("ram", "shyam"))
```

```python
print(sentence.count("ram"))
```

### 4. Check with `in`

```python
if "momo" in sentence:
    print("Someone likes momo!")
```

### 5. `split()` and `join()`
- `split()` cuts text into a **list** of words.
- `join()` joins a list back into text.

```python
school = "Ram Shah Secondary School"
```

```python
words = school.split()
```

```python
print(words)
```

```python
print(len(words), "words")
```

```python
print(" | ".join(words))
```

```python
print("-".join(words))
```

## 📌 Quick Recap
- `text[0]` first letter, `len(text)` length, `text[::-1]` reverse.
- `upper`, `lower`, `title`, `replace`, `count`.
- `split()` → list, `" ".join(list)` → text.
