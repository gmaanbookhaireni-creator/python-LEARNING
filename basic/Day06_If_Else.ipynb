# Day 6: Making Decisions (if / else) 🚦

**Today you will learn:**
- Comparing values
- `if`, `else` and `elif`

### 1. Comparing values
The answer is always `True` or `False`.

| Sign | Meaning |
|---|---|
| `>` `<` | greater / smaller |
| `>=` `<=` | greater or equal / smaller or equal |
| `==` | equal |
| `!=` | not equal |

```python
print(10 > 5)
```

```python
print(10 < 5)
```

```python
print(5 == 5)
```

```python
print(5 != 3)
```

⚠️ `=` puts a value in a variable. `==` checks if two values are equal.

```python
print(10 <= 15 <= 100)
```

### 2. `if`
The code under `if` runs only when the condition is `True`.
Don't forget the colon `:` and the **4 spaces** (indent).

```python
age = 18
```

```python
if age >= 18:
    print("You can vote!")
```

### 3. `if` + `else`
`else` runs when the condition is `False`.

```python
number = 7
```

```python
if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

### 4. `elif` (else if)
Use `elif` to check more conditions, one by one.

```python
marks = 72
```

```python
if marks >= 90:
    print("Grade A+")
elif marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

### 5. With user input

```python
password = input("Enter password: ")
```

```python
if password == "python123":
    print("Welcome!")
else:
    print("Wrong password")
```

### 6. Mini project: Calculator

```python
num1 = float(input("Enter first number: "))
```

```python
num2 = float(input("Enter second number: "))
```

```python
op = input("Enter operation (+ - * /): ")
```

```python
if op == "+":
    print(f"Result: {num1 + num2}")
elif op == "-":
    print(f"Result: {num1 - num2}")
elif op == "*":
    print(f"Result: {num1 * num2}")
elif op == "/":
    if num2 == 0:
        print("Error: cannot divide by zero!")
    else:
        print(f"Result: {num1 / num2}")
else:
    print("Error: invalid operation!")
```

## 📌 Quick Recap
- Comparisons give `True` or `False`.
- `if` → `elif` → `else`. Only **one** block runs.
- Always use `:` and indent with 4 spaces.
