# Day 9: While Loops, break and continue ⏳

**Today you will learn:**
- `while` loops
- Stopping a loop with `break`
- Skipping a step with `continue`

### 1. `while` loop
Repeats **while** the condition is `True`.

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

⚠️ If the condition never becomes `False`, the loop runs forever. Press the **■ Stop** button to stop it.

### 2. `break` — stop the loop

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

### 3. `continue` — skip this round

```python
for i in range(10):
    if i >= 3 and i <= 6:
        continue
    print(i)
```

### 4. Keep asking until correct
`while True` loops forever, so we use `break` to get out.

```python
while True:
    password = input("Enter password: ")
    if password == "abcd":
        print("Welcome!")
        break
    print("Incorrect password, try again.")
```

### 5. Game: guess the number
`random.randint(1, 10)` picks a secret number. (You met `random` on Day 3.)

```python
import random
```

```python
secret = random.randint(1, 10)
guess = 0
tries = 0

while guess != secret:
    guess = int(input("Guess a number (1-10): "))
    tries += 1

print(f"Correct! You found it in {tries} tries.")
```

## 📌 Quick Recap
- `while condition:` repeats until the condition is `False`.
- `break` stops the loop.
- `continue` skips to the next round.
