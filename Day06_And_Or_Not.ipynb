# Day 6: and, or, not 🔀

**Today you will learn:**
- Checking two conditions with `and` / `or`
- Flipping a condition with `not`

### 1. `and`
`True` only if **both** sides are `True`.

```python
print(True and True)
```

```python
print(True and False)
```

```python
marks = 75
```

```python
if marks >= 40 and marks <= 100:
    print("Pass")
```

### 2. `or`
`True` if **at least one** side is `True`.

```python
print(True or False)
```

```python
print(False or False)
```

```python
day = "Saturday"
```

```python
if day == "Saturday" or day == "Sunday":
    print("Holiday!")
else:
    print("School day")
```

### 3. `not`
`not` flips `True` to `False` and `False` to `True`.

```python
is_raining = False
```

```python
if not is_raining:
    print("Let's play football!")
```

### 4. Example: smallest of three numbers

```python
num1 = int(input("Number 1: "))
```

```python
num2 = int(input("Number 2: "))
```

```python
num3 = int(input("Number 3: "))
```

```python
if num1 <= num2 and num1 <= num3:
    print(num1, "is the smallest")
elif num2 <= num1 and num2 <= num3:
    print(num2, "is the smallest")
else:
    print(num3, "is the smallest")
```

## 📌 Quick Recap
- `and` → both must be `True`.
- `or` → one is enough.
- `not` → opposite.
