## 3. Age Calculator with Dates 🎂
**Field:** Everyday apps: calendars, banking, reminders

```python
from datetime import date

year = int(input("Birth year (like 2011): "))
month = int(input("Birth month (1-12): "))
day = int(input("Birth day (1-31): "))

days = (date.today() - date(year, month, day)).days

print()
print(f"🎉 You have been alive for {days:,} days")
print(f"⏰ That is {days * 24:,} hours")
print(f"⏱️ That is {days * 24 * 60 * 60:,} seconds")
print(f"❤️ Your heart has beaten about {days * 24 * 60 * 72:,} times!")
```

## 4. Secret Codes (Encryption) 🕵️
**Field:** Cybersecurity: passwords, WhatsApp messages, online banking

```python
def secret_code(text, shift):
    result = ""
    for ch in text:
        if ch.isalpha():
            start = ord("A") if ch.isupper() else ord("a")
            result += chr((ord(ch) - start + shift) % 26 + start)
        else:
            result += ch
    return result

message = input("Type a secret message: ")

hidden = secret_code(message, 3)
print("🔒 Secret code :", hidden)
print("🔓 Decoded back:", secret_code(hidden, -3))
```

## 5. Drawing with a Math Formula ❤️
**Field:** Computer graphics: shapes in games and animation are made with math

```python
name = input("Type your name: ")

for y in range(12, -11, -1):
    line = ""
    for x in range(-30, 30):
        a, b = x * 0.05, y * 0.1
        if (a * a + b * b - 1) ** 3 - a * a * b ** 3 <= 0:
            line += name[(x - y) % len(name)]
        else:
            line += " "
    print(line)
```

## 6. Rock, Paper, Scissors Game ✊✋✌️
**Field:** Game development: computer players and game rules

```python
import random

choices = ["rock", "paper", "scissors"]
beats = {"rock": "scissors", "paper": "rock", "scissors": "paper"}
emoji = {"rock": "✊", "paper": "✋", "scissors": "✌️"}
you_score, computer_score = 0, 0

for round_number in range(1, 4):
    you = input(f"Round {round_number} - rock, paper or scissors? ").lower().strip()
    if you not in choices:
        print("Oops! Type rock, paper or scissors.")
        continue
    computer = random.choice(choices)
    print(f"You {emoji[you]}  vs  Computer {emoji[computer]}")
    if you == computer:
        print("Draw! 🤝")
    elif beats[you] == computer:
        print("You win this round! 🎉")
        you_score += 1
    else:
        print("Computer wins this round! 🤖")
        computer_score += 1

print(f"\nFinal score - You: {you_score}, Computer: {computer_score}")
```

## 7. Chatbot 🤖
**Field:** Artificial Intelligence: ChatGPT, Claude, customer help bots

```python
import random
from datetime import datetime

jokes = [
    "Why did the computer go to the doctor? Because it had a virus! 🤒",
    "Why was the math book sad? It had too many problems! 📕",
    "What do you call a snake that writes code? A Python! 🐍",
]

print("🤖 Sathi: Namaste! I am Sathi, your robot friend. Type 'bye' to stop.")

while True:
    text = input("You: ").lower()
    words = text.split()
    if "bye" in text:
        print("🤖 Sathi: Bye bye! See you in Python class! 👋")
        break
    elif "hello" in words or "hi" in words or "namaste" in words:
        print("🤖 Sathi: Hello! How are you today? 😊")
    elif "name" in text:
        print("🤖 Sathi: My name is Sathi. I am made of Python code!")
    elif "how are you" in text:
        print("🤖 Sathi: I am great! Computers never get tired. 💪")
    elif "joke" in text:
        print("🤖 Sathi:", random.choice(jokes))
    elif "time" in text:
        print("🤖 Sathi: It is", datetime.now().strftime("%I:%M %p"), "⏰")
    elif "good" in text or "fine" in text:
        print("🤖 Sathi: Nice to hear that! 🎉")
    else:
        print("🤖 Sathi:", random.choice(["Interesting! Tell me more.", "Hmm, I am still learning. 🤔", "Wow! 😮"]))
```

## 8. Text to Speech 🔊
**Field:** Voice assistants: Siri, Alexa, Google Assistant

```python
import subprocess
import sys

text = input("What should I say? ")

if sys.platform == "darwin":                  # Mac
    subprocess.run(["say", text])
else:                                          # Windows / Linux
    try:
        import pyttsx3
        engine = pyttsx3.init()
        engine.say(text)
        engine.runAndWait()
    except ImportError:
        print("Run %pip install pyttsx3 first, then try again.")
```

## 9. Turtle Drawing 🐢
**Field:** Robotics: robots and drones also move with commands like forward and turn

```python
import turtle

screen = turtle.Screen()
screen.bgcolor("black")

t = turtle.Turtle()
t.speed(0)
t.hideturtle()
t.pensize(2)

colors = ["red", "orange", "yellow", "green", "cyan", "blue", "purple", "magenta"]

for i in range(180):
    t.pencolor(colors[i % len(colors)])
    t.forward(i * 2)
    t.left(59)
```

## 11. Fractals (Mandelbrot Set) 🌀
**Field:** Movies and science: computer-made mountains, clouds and nature patterns

```python
import numpy as np
import matplotlib.pyplot as plt

def mandelbrot(center_x, center_y, zoom, steps=300):
    size = 1.5 / zoom
    x = np.linspace(center_x - size * 1.5, center_x + size * 1.5, 600)
    y = np.linspace(center_y - size, center_y + size, 400)
    c = x[np.newaxis, :] + 1j * y[:, np.newaxis]
    z = np.zeros_like(c)
    count = np.zeros(c.shape)
    for i in range(steps):
        inside = np.abs(z) <= 2
        z[inside] = z[inside] ** 2 + c[inside]
        count[inside] += 1
    count[count == steps] = 0          # points that never escape are black
    return np.log(count + 1)

fig, boxes = plt.subplots(1, 3, figsize=(18, 6), facecolor="black")
for box, zoom in zip(boxes, [1, 30, 900]):
    box.imshow(mandelbrot(-0.7436, 0.1318, zoom), cmap="twilight_shifted", origin="lower")
    box.set_title(f"Zoom x{zoom}", color="white", fontsize=16)
    box.axis("off")
plt.show()
```

## 12. Data Chart: Nepal's Population 📈
**Field:** Data science: news charts, business reports, government planning

```python
import matplotlib.pyplot as plt

years = [1911, 1952, 1971, 1981, 1991, 2001, 2011, 2021]
population = [5.6, 8.3, 11.6, 15.0, 18.5, 23.2, 26.5, 29.2]   # in millions

plt.figure(figsize=(10, 5))
plt.plot(years, population, marker="o", color="crimson", linewidth=3)
plt.fill_between(years, population, alpha=0.2, color="crimson")
plt.title("Population of Nepal (Census)", fontsize=16)
plt.xlabel("Year")
plt.ylabel("People (millions)")
plt.grid(alpha=0.3)
plt.show()
```
