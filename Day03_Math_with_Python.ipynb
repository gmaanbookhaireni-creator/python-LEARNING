### 2. More operators
| Operator | Meaning | Example | Answer |
|---|---|---|---|
| `//` | divide, no decimal | `17 // 5` | `3` |
| `%` | remainder | `17 % 5` | `2` |
| `**` | power | `2 ** 3` | `8` |

```python
print(17 // 5)
```

```python
print(17 % 5)
```

```python
print(2 ** 3)
```

### 3. Order of math
Python follows the same rule as your math class. Brackets first, then `*` `/`, then `+` `-`.

```python
print(2 + 3 * 4)
```

```python
print((2 + 3) * 4)
```

### 4. Math with variables

```python
price = 25
```

```python
quantity = 4
```

```python
total = price * quantity
```

```python
print("Total:", total)
```

### 5. Updating a variable
`score += 5` is a short way to write `score = score + 5`.

```python
score = 10
```

```python
score += 5
```

```python
print(score)
```

### 6. `round()` and `abs()`
- `round(number, digits)` rounds a decimal.
- `abs(number)` removes the minus sign.

```python
print(round(5.656565, 2))
```

```python
print(abs(-5))
```

## 📌 Quick Recap
- `+ - * /` work like normal math.
- `//` no decimal, `%` remainder, `**` power.
- `x += 1` means `x = x + 1`.
