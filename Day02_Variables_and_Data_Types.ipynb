# Day 2: Variables and Data Types 📦

**Today you will learn:**
- How to store values in **variables**
- The 4 basic data types
- How to check a type with `type()`

### 1. What is a variable?
A variable is a **box with a name** that stores a value. `=` puts the value in the box.

````python
name = "Sita"
````

````python
print(name)
````

````python
age = 14
````

````python
print(age)
````

### 2. Changing a variable
Put a new value in the same box. The old value is gone.

````python
age = 15
````

````python
print(age)
````

### 3. Using many variables

````python
name = "Sita"
````

````python
age = 14
````

````python
school = "Ram Shah School"
````

````python
print(name, age, school)
````

### 4. Data types
| Type | Meaning | Example |
|---|---|---|
| `str` | text | `"Ram"` |
| `int` | whole number | `14` |
| `float` | decimal number | `5.5` |
| `bool` | True or False | `True` |

````python
name = "Ram"
````

````python
age = 14
````

````python
height = 5.5
````

````python
is_student = True
````

````python
print(type(name))
````

````python
print(type(age))
````

````python
print(type(height))
````

````python
print(type(is_student))
````

### 5. `"5"` is not the same as `5`
Anything inside quotes is text (`str`).

````python
print(type("5"))
````

````python
print(type(5))
````

### 6. Naming rules
- ✅ Good: `name`, `student_name`, `marks1`
- ❌ Bad: `1name` (starts with number), `student name` (has space)
- Names are case sensitive: `Age` and `age` are different.

## 📌 Quick Recap
- `variable = value` stores a value.
- Types: `str` (text), `int` (whole), `float` (decimal), `bool` (True/False).
- `type(x)` tells the type of `x`.
