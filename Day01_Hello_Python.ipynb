### 1. Your first program
`print()` shows something on the screen. Text goes inside quotes `" "`.

````python
print("Hello, World!")
````

````python
print("My name is Ram")
````

````python
print("I study in class 9")
````

### 2. Printing numbers
Numbers do **not** need quotes. Python can also do math inside `print()`.

````python
print(10)
````

````python
print(5 + 3)
````

### 3. Text vs number
With quotes it is just text. Without quotes Python calculates it.

````python
print("5 + 3")
print(5 + 3)
````

### 4. Printing many things together
Use commas `,` to print many things in one line.

````python
print("Ram", "is", 14, "years old")
````

### 5. Empty line
An empty `print()` prints a blank line.

````python
print("Good morning")
print()
print("Let's learn Python")
````

### 6. Comments
Anything after `#` is a **comment**. Python ignores it. It is a note for humans.

````python
# This line prints a greeting
print("Namaste!")  # comments can also go at the end of a line
````
