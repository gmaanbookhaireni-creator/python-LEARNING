# Day 4: Talking to the User 💬

**Today you will learn:**
- How to take input with `input()`
- How to change text into numbers
- How to print nicely with **f-strings**

### 1. `input()`
`input()` waits for the user to type something. Type and press **Enter**.

```python
name = input("What is your name? ")
```

```python
print("Hello", name)
```

### 2. Input is always text
Even if you type a number, `input()` gives a `str`.

```python
age = input("How old are you? ")
```

```python
print(type(age))
```

### 3. Change text to a number
- `int()` → whole number
- `float()` → decimal number

```python
age = int(input("How old are you? "))
```

```python
print("Next year you will be", age + 1)
```

### 4. f-strings
Put `f` before the quotes. Then put variables inside `{ }`.

```python
name = "Sita"
```

```python
age = 14
```

```python
print(f"My name is {name} and I am {age} years old.")
```

```python
print(f"In 5 years I will be {age + 5}.")
```

### 5. Small program: add two numbers

```python
a = int(input("First number: "))
```

```python
b = int(input("Second number: "))
```

```python
print(f"{a} + {b} = {a + b}")
```

### 6. Small program: write an email

```python
sender = input("From: ")
```

```python
receiver = input("To: ")
```

```python
subject = input("Subject: ")
```

```python
message = input("Message: ")
```

```python
print()
```

```python
print(f"From    : {sender}")
```

```python
print(f"To      : {receiver}")
```

```python
print(f"Subject : {subject}")
```

```python
print(f"Message : {message}")
```

## 📌 Quick Recap
- `input("question")` gets text from the user.
- Use `int()` or `float()` to change it into a number.
- `f"Hello {name}"` puts variables inside text.
